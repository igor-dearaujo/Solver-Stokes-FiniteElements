````python
### QUESTION 1.5

def elem_mat_K(S1, S2, S3, tol=1e-8):
    """Calcule la matrice de rigidité élémentaire en P^1 Lagrange.

    NOTE (1) le calcul est exact (pas de condensation de masse)
         (2) calcul direct à partir des formules données par les
         coordonnées barycentriques

    Parameters
    ----------
    S1, S2, S3 : les 2 coordonnées des 3 sommets du triangle
        (vecteurs reels de taille (1, 2))

    Returns
    -------
    matrice de rigidité elementaire (matrice 3x3)
    """

    # preliminaires, pour faciliter la lecture:
    x1, y1 = S1
    x2, y2 = S2
    x3, y3 = S3

    # D est, au signe près, deux fois l'aire du triangle
    D = (x2 - x1) * (y3 - y1) - (y2 - y1) * (x3 - x1)

    if abs(D) <= tol:
        raise ValueError("l'aire d'un triangle est nulle !!!")

    # calcul de la matrice de rigidité
    # Réponse 1.5 - début
    K_elem = np.zeros((3, 3))
    
    v1 = (y2 - y3, x3 - x2)
    v2 = (y3 - y1, x1 - x3)
    v3 = (y1 - y2, x2 - x1)

    V = [v1, v2, v3]

    K_elem = (1.0 / (2.0 * abs(D))) * ((V) @ np.transpose(V))
    # fin

    return K_elem
````

````python
#MATRICE DE MASSE
def elem_mat_M(S1, S2, S3, tol=1e-8):
    """Calcule la matrice de masse élémentaire en P^1 Lagrange.

    NOTE (1) le calcul est exact (pas de condensation de masse)
         (2) calcul direct à partir des formules données par les
         coordonnées barycentriques

    Parameters
    ----------
    S1, S2, S3 : les 2 coordonnées des 3 sommets du triangle
        (vecteurs reels de taille (1, 2))

    Returns
    -------
    matrice de masse elementaire (matrice 3x3)
    """

    # preliminaires, pour faciliter la lecture:
    x1, y1 = S1
    x2, y2 = S2
    x3, y3 = S3

    # D est, au signe près, deux fois l'aire du triangle
    D = (x2 - x1) * (y3 - y1) - (y2 - y1) * (x3 - x1)

    if abs(D) <= tol:
        raise ValueError("l'aire d'un triangle est nulle !!!")

    # calcul de la matrice de masse
    # Réponse 1.5 - début
    M_elem = ((abs(D) / 24.0) * np.array([[2, 1, 1], [1, 2, 1], [1, 1, 2]]))
    # fin

    return M_elem
````

````python
#QUESTION 1.8
def mat_KM(mesh):
    """Calcule les matrices de rigidité et de masse en P^1 Lagrange.

    Parameters
    ----------
    mesh : Mesh
        le maillage

    Returns
    -------
    K : la matrice de rigidité (matrice sparse de taille nb_nodes x nb_nodes)
    M : la matrice de masse (matrice sparse de taille nb_nodes x nb_nodes)
    """
    K = sp.dok_array((mesh.nb_nodes, mesh.nb_nodes), dtype=np.float64)
    M = sp.dok_array((mesh.nb_nodes, mesh.nb_nodes), dtype=np.float64)

    for k in range(mesh.nb_elements("Omega", "triangle")): 
        # Coordonnées des sommets du triangle
        # Réponse 1.8 - début
        tri = mesh.domains["Omega"]["triangle"][k]
        i1, i2, i3 = tri[0], tri[1], tri[2]
        S1 = mesh.nodes[i1]
        S2 = mesh.nodes[i2]
        S3 = mesh.nodes[i3]

        # calcul des matrices elementaires du triangle
        K_elem = elem_mat_K(S1, S2, S3)
        M_elem = elem_mat_M(S1, S2, S3)

        # On fait l'assemblage de la matrice globale et du second membre
        for i in range(3):
            I = tri[i]
            for j in range(3):
                J = tri[j]
                K[I, J] += K_elem[i, j]
                M[I, J] += M_elem[i, j]
        # fin

    return (K, M)
````

````python
#QUESTION 1.10
def vec_L(fct, mass_matrix, nodes):
    """Calcule le second membre L.

    Parameters
    ----------
    fct : la fonction `fct` prend comme argument les nœuds du maillage
          (tableau de taille `(nb_nodes, 2)`)
          et renvoie un vecteur de taille `nb_nodes` qui contient les
          valeurs de `fct` en chaque nœud. Un
          exemple de fonction `fct` est donné ci-dessous :
      ```def f(nodes):
            x, y = nodes[:, 0], nodes[:, 1]
            return np.cos(x) * np.exp(y)
      ```

    mass_matrix : matrice de masse (matrice sparse de taille nb_nodes x nb_nodes)

    nodes : les coordonnées des nœuds du maillage (matrice de taille nb_nodes x 2)

    Returns
    -------
    vecteur L (vecteur de taille nb_nodes x 1)
    """

    # Réponse 1.10 - début
    L = mass_matrix @ fct(nodes)
    return L
    # fin
````

Comment évolue cette erreur en fonction de $h$ dans le cas de la première géométrie ?

````python
def u_function(nodes):
    # Function that defines the function
    x, y = nodes[:, 0], nodes[:, 1]
    return np.cos(3 * np.pi * x) * np.cos(2 * np.pi * y)

def interpolate_in_nodes(fct, nodes):
    return fct(nodes)

U_exact = interpolate_in_nodes(u_function, mesh.nodes)

def calculate_L2_error(M, U, U_exact):
    """
    Calculates the finite element L2 norm error ||U - U_exact||_L2 in a given mesh
    """
    diff = U - U_exact
    error_sqrd = diff.T @ M @ diff
    norme_U_sqrd = U.T @ M @ U
    return np.sqrt(abs(float(np.squeeze(error_sqrd))) / abs(float(np.squeeze(norme_U_sqrd))))

l2_err = calculate_L2_error(M, U, U_exact)
print(f"Norme L2 de l'erreur : {l2_err}")

def calculate_H1_error(K, U, U_exact):
    """
    Calculates the finite element H1 semi-norm error ||U - U_exact||_H1
     = ||grad(U) - grad(U_exact)||_L2 in a given mesh
    """
    diff = U - U_exact
    error_sqrd = diff.T @ K @ diff
    norme_U_sqrd = U.T @ K @ U
    return np.sqrt(abs(float(np.squeeze(error_sqrd))) / abs(float(np.squeeze(norme_U_sqrd))))


    
````

````python
# Study of multiple passes:

def error_study(h_list, f, U_fct):
    
    L2_errors = {}
    H1_errors = {}

    for h in h_list:
        # Define Mesh
        mesh = src.rectangle((0, 0), (1, 1), h)
        
        # Define mass & rigidty matrices
        K, M = mat_KM(mesh)
        
        # Define right-hand side
        L = vec_L(f, M, mesh.nodes)

        # Calc U
        U = src.sparse_solve(K + M, L)

        # Calc U_exact
        U_exact = interpolate_in_nodes(U_fct, mesh.nodes)

        # Calc L2 error
        L2_error = calculate_L2_error(M, U, U_exact)
        H1_error = calculate_H1_error(K, U, U_exact)

        L2_errors[h] = L2_error 
        H1_errors[h] = H1_error
    
    h_L2 = np.array(list(L2_errors.keys()), dtype=float)
    err_L2 = np.array(list(L2_errors.values()), dtype=float)

    h_H1 = np.array(list(H1_errors.keys()), dtype=float)
    err_H1 = np.array(list(H1_errors.values()), dtype=float)

    return L2_errors, H1_errors

h = np.array([0.025, 0.05, 0.10, 0.20])
L2_dict, H1_dict = error_study(h, f, u_function)
L2_errors, H1_errors = list(L2_dict.values()), list(H1_dict.values())


print([L2_errors, H1_errors])
````

````python
fitL2_func = np.poly1d(np.polyfit(np.log(1/h), np.log(L2_errors), 1))
fitH1_func = np.poly1d(np.polyfit(np.log(1/h), np.log(H1_errors), 1))

fig, ax = plt.subplots() 

ax.loglog(1/h, L2_errors, 'or', label='L2 Error', linewidth=2)
ax.loglog(1/h, np.exp(fitL2_func(np.log(1/h))), '--r', label=f'Regression L2, pente: {fitL2_func[1]:.2f}', linewidth=2 )

ax.loglog(1/h, H1_errors, 'ob', label= 'H1 Error', linewidth=2)
ax.loglog(1/h, np.exp(fitH1_func(np.log(1/h))), '--b', label=f'Regression H1, pente: {fitH1_func[1]:.2f}', linewidth=2 )

ax.set_xlabel('Mesh refinement [log(1/h)]')
ax.set_ylabel('Error')
ax.grid(visible=True)
ax.set_title('Error study')
ax.legend()
plt.savefig('images/fig_errorstudy_elementsP1.png')

plt.show()
````

Comment évolue cette erreur en fonction de $h$ dans le cas de la première géométrie ?

<div class="alert alert-block alert-info">
<strong>Comment effectuer une étude complète de l'erreur</strong>

Tracer un graphe __en échelle log-log__ de l'erreur normalisée en norme $L^2$, puis en semi-norme $H^1$, en fonction du réciproque du pas du maillage $h$.
Plus précisément :
- En abscisse, on a $\log(1/h)$ et en ordonnée, on a $\log(\lVert u-u_h \rVert_{L^2(\Omega)}) / \lVert u\rVert_{L^2(\Omega)})$, puis $\log(\lvert u-u_h \rvert_{H^1(\Omega)}) / \lvert u\rvert_{H^1(\Omega)})$.
- Choisir au moins $3$ ou $4$ valeurs différentes de $h$ : $h_1 > h_2 > h_3 > h_4$ (souvent, on choisit $h_{i+1} = \alpha h_i$ pour que les points du graphe soient bien espacés en échelle $\log$).
- Effectuer une régression linéaire (par exemple avec la fonction `np.polyfit`) pour calculer les coefficients de la droite de régression.
- Tracer les courbes d'erreur (mettre bien en évidence les points sur les courbes), ainsi que les droites de régression (avec un autre style, par exemple en pointillé).
